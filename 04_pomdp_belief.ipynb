```python
import sys
sys.path.append("..")

import numpy as np
from pomdp import make_tiger_pomdp, TabularPOMDP
```

```python
pomdp = make_tiger_pomdp()
print("States:", pomdp.states)
print("Actions:", pomdp.actions)
print("Observations:", pomdp.observations)
```

```python
# Start with uniform belief
belief = np.array([0.5, 0.5])
print("Initial belief:")
pomdp.print_belief(belief)
```

```python
print("\nAction: listen, Observation: hear-left")
new_belief = pomdp.belief_update(belief, action="listen", observation="hear-left")
print("Updated belief:")
pomdp.print_belief(new_belief)

print("\nExpected: roughly [0.85, 0.15] (more belief that tiger is on left)")
```

```python
print("\nAgain listen + hear-left")
newer_belief = pomdp.belief_update(new_belief, action="listen", observation="hear-left")
print("Updated belief:")
pomdp.print_belief(newer_belief)

print("\nBelief on tiger-left should go higher (around 0.97)")
```
